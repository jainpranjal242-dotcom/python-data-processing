## Python Data Processing

Process structured student records and generate useful summary statistics using Python lists, dictionaries, functions, loops, and built-in statistical operations.

```python
records = [
    {'id': 101, 'name': 'Aarav', 'department': 'CSE', 'age': 20, 'marks': 86, 'attendance': 92},
    {'id': 102, 'name': 'Diya', 'department': 'AI-ML', 'age': 19, 'marks': 91, 'attendance': 95},
    {'id': 103, 'name': 'Kabir', 'department': 'ECE', 'age': 21, 'marks': 74, 'attendance': 81},
    {'id': 104, 'name': 'Ananya', 'department': 'CSE', 'age': 20, 'marks': 88, 'attendance': 89},
    {'id': 105, 'name': 'Rohan', 'department': 'AI-ML', 'age': 19, 'marks': 79, 'attendance': 84},
    {'id': 106, 'name': 'Meera', 'department': 'CSE', 'age': 21, 'marks': 95, 'attendance': 97},
    {'id': 107, 'name': 'Vivaan', 'department': 'ECE', 'age': 20, 'marks': 68, 'attendance': 76},
    {'id': 108, 'name': 'Ishita', 'department': 'AI-ML', 'age': 20, 'marks': 84, 'attendance': 91},
    {'id': 109, 'name': 'Arjun', 'department': 'CSE', 'age': 19, 'marks': 73, 'attendance': 80},
    {'id': 110, 'name': 'Sana', 'department': 'ECE', 'age': 21, 'marks': 90, 'attendance': 93}
]
```

### 1. Basic dataset information

```python
def dataset_info(data):
    print('Total records:', len(data))
    print('Departments:', sorted({student['department'] for student in data}))

dataset_info(records)
```

### 2. Summary statistics

```python
def summary_statistics(data):
    marks = [student['marks'] for student in data]
    attendance = [student['attendance'] for student in data]
    ages = [student['age'] for student in data]
    return {
        'average_marks': sum(marks) / len(marks),
        'highest_marks': max(marks),
        'lowest_marks': min(marks),
        'average_attendance': sum(attendance) / len(attendance),
        'average_age': sum(ages) / len(ages)
    }

stats = summary_statistics(records)
for key, value in stats.items():
    print(f'{key}: {value:.2f}')
```

### 3. Filter high-performing students

```python
def high_performers(data, threshold=85):
    return [student for student in data if student['marks'] >= threshold]

for student in high_performers(records):
    print(student['name'], '-', student['marks'])
```

### 4. Department-wise statistics

```python
def department_statistics(data):
    result = {}
    for student in data:
        dept = student['department']
        result.setdefault(dept, []).append(student['marks'])
    return {dept: {'count': len(marks), 'average_marks': sum(marks)/len(marks)}
            for dept, marks in result.items()}

department_stats = department_statistics(records)
department_stats
```

### 5. Processed record examples

```python
def process_record(student):
    result = student.copy()
    result['performance'] = 'Excellent' if student['marks'] >= 90 else ('Good' if student['marks'] >= 75 else 'Needs Improvement')
    result['attendance_status'] = 'Eligible' if student['attendance'] >= 75 else 'Low Attendance'
    return result

processed_records = [process_record(student) for student in records]
for student in processed_records[:5]:
    print(student)
```
